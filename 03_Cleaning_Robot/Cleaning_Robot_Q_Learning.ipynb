# Intelligent Cleaning Robot using Q-Learning

**Objective:** Simulate a room-cleaning agent that learns useful actions through rewards and penalties.

**AI concept:** Reinforcement Learning.


```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
rows, cols = 5, 5
environment = np.zeros((rows, cols), dtype=int)

dirty_cells = [(0, 3), (1, 1), (2, 4), (3, 2), (4, 0)]
for cell in dirty_cells:
    environment[cell] = 1

environment
```

```python
actions = ["up", "down", "left", "right", "clean"]
num_states = rows * cols
num_actions = len(actions)

q_table = np.zeros((num_states, num_actions))

def get_state(row, col):
    return row * cols + col

def move_robot(row, col, action):
    if action == 0:
        row = max(0, row - 1)
    elif action == 1:
        row = min(rows - 1, row + 1)
    elif action == 2:
        col = max(0, col - 1)
    elif action == 3:
        col = min(cols - 1, col + 1)
    return row, col
```

```python
learning_rate = 0.1
discount_factor = 0.9
epsilon = 0.2
episodes = 1000

episode_rewards = []

for episode in range(episodes):
    room = environment.copy()
    row = random.randint(0, rows - 1)
    col = random.randint(0, cols - 1)
    total_reward = 0

    for step in range(200):
        state = get_state(row, col)

        if random.random() < epsilon:
            action = random.randint(0, num_actions - 1)
        else:
            action = int(np.argmax(q_table[state]))

        reward = -1

        if action == 4:
            if room[row, col] == 1:
                reward = 10
                room[row, col] = 0
            else:
                reward = -2
            next_row, next_col = row, col
        else:
            next_row, next_col = move_robot(row, col, action)

        next_state = get_state(next_row, next_col)

        q_table[state, action] += learning_rate * (
            reward
            + discount_factor * np.max(q_table[next_state])
            - q_table[state, action]
        )

        row, col = next_row, next_col
        total_reward += reward

        if np.sum(room) == 0:
            total_reward += 20
            break

    episode_rewards.append(total_reward)
```

```python
plt.figure(figsize=(10, 5))
plt.plot(episode_rewards)
plt.xlabel("Episode")
plt.ylabel("Total Reward")
plt.title("Cleaning Robot Learning Progress")
plt.show()
```

```python
window = 50
moving_avg = np.convolve(
    episode_rewards,
    np.ones(window) / window,
    mode="valid"
)

plt.figure(figsize=(10, 5))
plt.plot(moving_avg)
plt.xlabel("Episode")
plt.ylabel("Average Reward")
plt.title("Moving Average Reward")
plt.show()
```

## How to explain the system

- **Agent:** cleaning robot
- **Environment:** 5×5 room
- **State:** robot position
- **Actions:** up, down, left, right, clean
- **Positive reward:** cleaning dirt
- **Negative reward:** unnecessary movement or cleaning a clean cell
- **Q-Learning:** updates action values so the robot gradually favors actions associated with higher long-term reward
